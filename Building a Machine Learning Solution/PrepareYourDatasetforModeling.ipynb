# Credit Card Fraud Detection Preprocessing Pipeline

# Step 1: Load and Explore the Dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

# Load dataset
df = pd.read_csv('creditcard.csv')

# Initial exploration
print("Shape:", df.shape)
print(df.head())
print(df.info())
print(df.describe())
print("Class distribution:\n", df['Class'].value_counts(normalize=True))

# Step 2: Handle Missing Data
# Check for missing values
print("Missing values:\n", df.isnull().sum())  # No missing values expected

# Step 3: Normalize or Scale Features
# Scale 'Amount' and 'Time' features
scaler = StandardScaler()
df['Amount_scaled'] = scaler.fit_transform(df[['Amount']])
df['Time_scaled'] = scaler.fit_transform(df[['Time']])

# Drop original 'Amount' and 'Time'
df.drop(['Amount', 'Time'], axis=1, inplace=True)

# Step 4: Encode Categorical Variables
# No categorical variables in this dataset — all are numerical

# Step 5: Detect and Treat Outliers
# Visualize outliers in 'Amount_scaled'
sns.boxplot(x='Class', y='Amount_scaled', data=df)
plt.title("Boxplot of Transaction Amount by Class")
plt.show()

# Decision: Retain all values as outliers may be informative for fraud

# Step 6: Select Features
# All features are anonymized and assumed relevant
X = df.drop('Class', axis=1)
y = df['Class']

# Step 7: Prepare the Dataset for Modeling
# Stratified split to maintain class balance
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

print("Training set size:", X_train.shape)
print("Test set size:", X_test.shape)
